```python
def predict(i):
    return [(e.text, e.label_) for e in nlp(sentences[i]).ents if e.label_ in KEEP]

def lenient_hit(p, g):
    a, b = p[0].lower(), g[0].lower()
    return p[1] == g[1] and (a in b or b in a)

def scores(gold):
    tp_s = tp_lp = tp_lr = pred_n = gold_n = 0
    for i, g in gold.items():
        pred = predict(i)
        tp_s  += len(set(pred) & set(g))                              # strict: exact text + label
        tp_lp += sum(any(lenient_hit(p, x) for x in g) for p in pred)  # lenient, for precision
        tp_lr += sum(any(lenient_hit(p, x) for p in pred) for x in g)  # lenient, for recall
        pred_n += len(pred)
        gold_n += len(g)
    print("strict   precision", round(tp_s / max(pred_n, 1), 3), " recall", round(tp_s / max(gold_n, 1), 3))
    print("lenient  precision", round(tp_lp / max(pred_n, 1), 3), " recall", round(tp_lr / max(gold_n, 1), 3))
    print("predicted:", pred_n, " gold:", gold_n)

scores(gold)
```
